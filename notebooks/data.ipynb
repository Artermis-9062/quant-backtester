```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))


from quantlib.data import load_prices, check_prices, write_manifest, verify_manifest

spy = load_prices("SPY", "2010-01-01", "2024-12-31")   # first call downloads
print(spy.shape, spy.index.min().date(), spy.index.max().date())
print(spy.head(3)); print(spy.describe().loc[["min", "max"]])
print(check_prices(spy))          # expect []
write_manifest(["SPY"]); print(verify_manifest())   # True
```

This will format the table

```python
import pandas as pd

# Load the CSV
spy = pd.read_csv("data/SPY.csv", index_col=0, parse_dates=True)

# Format the display: 2 decimals for prices, 0 for volume
pd.options.display.float_format = "{:,.2f}".format

# Show it
print("SPY Data (formatted):")
print(spy.head(10))
print("\nSummary statistics:")
print(spy.describe())
```

```python
bad = spy.copy()
bad.iloc[100, bad.columns.get_loc("Close")] *= 10        # bad tick
print(check_prices(bad))
bad = spy.drop(spy.index[200:210])                        # missing two weeks
print(check_prices(bad))
bad = spy.copy(); bad.iloc[50, bad.columns.get_loc("High")] = 1.0
print(check_prices(bad))
```
